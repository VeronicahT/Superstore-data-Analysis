```python
with zipfile.ZipFile(r'C:\Users\Toror\Downloads\archive (1).zip') as z:
    with z.open('Superstore Sales Dataset.csv') as f:
        df = pd.read_csv(f, sep=';', encoding='latin1')

print(df.shape)
print(df.dtypes)
print(df.isnull().sum())
```

```python
df['Sales'] = (
    df['Sales']
    .str.strip()
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
    .astype(float)
    / 100
)
df['Sales'].head()
```

```python
suspect = df[df['Sales'] > 20000]
print(suspect[['Row ID', 'Product Name', 'Sales']])
print(f"\n{len(suspect)} suspect rows out of {len(df)}")
```

```python
# Rows where /100 produced an implausibly large value
suspect_mask = df['Sales'] > 20000

# These lost an extra 2 decimal places during corruption - divide by 100 again
df.loc[suspect_mask, 'Sales'] = df.loc[suspect_mask, 'Sales'] / 100

# Recheck
print(df[df['Sales'] > 20000])   # should now be empty or near-empty
print(df['Sales'].describe())
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)

print(df[['Order Date', 'Ship Date']].dtypes)
df[['Order Date', 'Ship Date']].head()
```

```python
df = df.drop(columns=['Month & Year Order'])
print(df.columns)
```

```python
df['Postal Code'] = df['Postal Code'].fillna(5401)
df['Postal Code'] = df['Postal Code'].astype(int).astype(str).str.zfill(5)

print(df['Postal Code'].dtype)
print(df['Postal Code'].isnull().sum())
df[df['City'] == 'Burlington'][['City', 'State', 'Postal Code']]
```

```python
# Checking duplicates

print(df.duplicated().sum())
print(df['Row ID'].duplicated().sum())
```

```python
# Save the file

print(df.info())
print(df.describe())

df.to_csv('superstore_cleaned.csv', index=False)

```
